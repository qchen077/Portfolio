## 8. Why Use Spatial Overlay Instead of Simple Spatial Join?

Zoning Boundaries and Existing Land Use are polygon datasets. A simple spatial join would only identify whether a zoning or land-use polygon intersects a census tract.

However, this analysis needs to know how much of each census tract is covered by each zoning or land-use category.

Therefore, this deliverable uses polygon overlay rather than simple spatial join.

The overlay process cuts zoning and land-use polygons by census tract boundaries. This allows the analysis to calculate the area of each zoning or land-use category inside each census tract.

The final tract-level variables are area shares, such as:

* share of tract area zoned residential
* share of tract area zoned apartment residential
* share of tract area zoned mixed-use or commercial
* share of tract area used for residential purposes
* share of tract area used for commercial purposes
* share of tract area used for institutional or industrial purposes

## 11. Modeling Direction

After EDA, the next modeling step is to test whether zoning and land-use variables add explanatory value to the tract-level crash count model.

The recommended model type remains Negative Binomial regression because the outcome is a crash count and the data are likely overdispersed.

The outcome should be:

`total_crashes`

The offset should be:

`log(area_km2 × n_years)`

Recommended model sequence:

### Model 1: Roadway Controls

Includes roadway network variables such as:

* road density
* intersection density
* signal density
* mean AADT
* major route density

### Model 2: Roadway Controls + Demographics

Adds tract-level demographic composition variables, such as:

* Black share
* Hispanic share

### Model 3: Roadway Controls + Demographics + Zoning

Adds selected zoning share variables.

### Model 4: Roadway Controls + Demographics + Existing Land Use

Adds selected existing land-use share variables.

### Model 5: Combined Built-Environment Model

Combines selected roadway, demographic, zoning, and land-use variables.

The purpose is to test whether built-environment context helps explain crash concentration beyond roadway exposure alone.

---

## 12. Important Modeling Caution

Zoning and land-use share variables can be highly correlated because area shares within a census tract may sum to approximately one. Therefore, not all categories should be included in the same model at once.

One category should be treated as the reference category, or only a small number of theoretically important categories should be included.

For example, the model may include:

* apartment residential zoning share
* mixed-use or commercial zoning share
* production/distribution/repair zoning share

while leaving low-density residential zoning as the implicit comparison category.

This helps reduce multicollinearity and improves interpretability.

<h2 style="color:#B00020; font-weight:700;">
What Zoning and Land Use Area Shares Represent
</h2>

Zoning and existing land use describe two different aspects of the built environment.

**Zoning** represents the legally regulated development or land-use category of a parcel or polygon. It describes what type of development or use is permitted or regulated in that area. Examples include Mixed-Use Zone, Residential Apartment Zone, Residential Zone, Residential Flat Zone, and Production, Distribution, and Repair Zone.

**Existing Land Use** represents how the land is currently being used. Examples may include residential, commercial, institutional, industrial, public or open space, and mixed use.

In this deliverable, zoning and land-use polygons are converted into census tract-level area shares. This does not mean that zoning itself is an area variable. Instead, zoning is a category assigned to each polygon, and area share measures how much of a census tract is covered by each category.

For example, if a census tract has a total area of 1 km² and the polygon overlay shows:

```text
Mixed-Use Zone                       0.30 km²
Residential Apartment Zone           0.20 km²
Residential Zone                     0.40 km²
Production, Distribution, Repair     0.10 km²
```

then the tract-level zoning shares are:

```text
Mixed-Use Zone share                  30%
Residential Apartment Zone share      20%
Residential Zone share                40%
PDR Zone share                        10%
```

This converts the original polygon-level zoning and land-use datasets into tract-level variables that can be merged with the existing crash, roadway, demographic, and major route exposure table.

In short:

```text
Zoning = legally allowed or regulated use
Existing Land Use = current actual use
Area share = percentage of a census tract covered by each category
```


## 4. Clean Category Fields

### Why Reproject to EPSG:26918?

The zoning, existing land use, and census tract layers are polygon datasets. To calculate area correctly, all spatial layers are reprojected to EPSG:26918, a meter-based projected coordinate system appropriate for Washington, DC.

In EPSG:4326, coordinates are stored as longitude and latitude in degrees, so polygon area calculations are not meaningful. In EPSG:26918, coordinates are measured in meters, so `geometry.area` returns square meters. Dividing by 1,000,000 converts square meters to square kilometers.

### Zoning vs. Existing Land Use

Zoning represents the legally regulated development or land-use category of a parcel or polygon. It describes what type of use is permitted or regulated in that area.

Existing Land Use represents how the land is currently being used.

These two concepts are related but not identical. A parcel may be zoned for mixed-use development, but its current land use may be residential, commercial, institutional, parking, or another actual use. Therefore, zoning captures the regulatory built environment, while existing land use captures the actual built environment.

Existing Land Use tells what the land currently is. Zoning tells what the land is legally allowed or regulated to become. Both may help explain crash concentration differently.

````python
tract_route_exposure_df = pd.read_csv(
    "../../../Data/tract_route_exposure_df.csv", dtype={"GEOID": str}
)
tract_route_df = tract_route_exposure_df.copy()

acs_demographics = gpd.read_file(
    "../../../Data/ACS_5-Year_Demographic_Characteristics_of_DC_Census_Tracts/ACS_5-Year_Demographic_Characteristics_of_DC_Census_Tracts.shp"
)

tract_poly = acs_demographics[["GEOID", "geometry"]].copy()
tract_poly["GEOID"] = tract_poly["GEOID"].astype(str).str.zfill(11)

# Measurement CRS for DC route length / area
# EPSG:26918: Best Use - local distance, length, area calculation around DC / East Coast
# EPSG 3857: web map display, Google maps, OpenStreetMap
target_crs = "EPSG:26918"

zoning = zoning.to_crs(target_crs)
tract_poly = tract_poly.to_crs(target_crs)
landuse = landuse.to_crs(target_crs)
````

````python
# Use broad zoning district category for first-stage analysis
zoning_field = "ZONE_DISTRICT"

# Keep only zoning category and geometry
zoning_small = zoning[[zoning_field, "geometry"]].copy()
zoning_small = zoning_small.rename(columns={zoning_field: "zoning_category"}) # rename colname: original name - ZONE-DISTRICT
zoning_small.head()
````

````python
# Clean category names for easier column names later
zoning_category_map = {
    "Mixed-Use Zone": "mixed_use",
    "Neighborhood Mixed-Use Zone": "neighborhood_mixed_use",
    "Residential Apartment Zone": "residential_apt",
    "Residential Zone": "residential",
    "Residential Flat Zone": "residential_flat",
    "Production, Distribution, and Repair Zone": "production_distribution_repair",
    "Special Purpose Zone": "special_purpose",
    "Downtown Zone": "downtown",
    "Unzoned": "unzoned"
}

zoning_small["zoning_category"] = zoning_small["zoning_category"].map(zoning_category_map)
````

````python
# Spatial overlay: cut zoning polygons by census tract boundaries
tract_zoning = gpd.overlay(
    tract_poly[["GEOID", "geometry"]],
    zoning_small,
    how="intersection"
)

# Calculate zoning area after intersection
# 某个 zoning category 在某个 tract 里面切出来的小块面积
tract_zoning["zoning_piece_area_km2"] = tract_zoning.geometry.area / 1_000_000 # squared meters -> squared km

tract_zoning.head()
````

````python
tract_zoning.shape
````

````python
# Groupby 后的 tract-category 总面积
# zoning_category_area_km2 = 某个 GEOID 里某一种 zoning category 的总面积
zoning_area = (
    tract_zoning
    .groupby(["GEOID", "zoning_category"], as_index=False)
    .agg(zoning_category_area_km2=("zoning_piece_area_km2", "sum"))
)

zoning_area.head()
````

````python
zoning_area.shape
````

### Why Convert Zoning Area from Long Format to Wide Format?

The `zoning_area` table is a tract-category-level table. Each row represents one census tract and one zoning category.

For example, the same `GEOID` may appear multiple times:

```text
GEOID        zoning_category        zoning_area_km2
11001000101  mixed_use              0.007
11001000101  residential            0.063
11001000101  residential_apt        0.049
11001000101  unzoned                0.085
```

This format is useful for checking zoning categories, but it should not be directly merged into the tract-level crash table. The main crash and route exposure table, tract_route_df, has one row per census tract. If zoning_area were merged directly, the same crash and roadway variables would be repeated multiple times for the same tract.

To avoid duplicating tract-level records, the zoning table is converted to wide format. In the wide table, each GEOID appears once, and each zoning category becomes its own column:
```text
GEOID        mixed_use   residential   residential_apt   unzoned
11001000101  0.007       0.063         0.049             0.085
```

This structure matches tract_route_df, where each row represents one census tract. After converting zoning areas into area-share variables, the zoning table can be safely merged with the crash, roadway, demographic, and major route exposure table.

```python
zoning_area_wide = zoning_area.pivot_table(
    index="GEOID",
    columns="zoning_category",
    values="zoning_category_area_km2",
    aggfunc="sum",
    fill_value=0
).reset_index()
```

````python
zoning_area_wide = zoning_area.pivot_table(
    index="GEOID",
    columns="zoning_category",
    values="zoning_category_area_km2",
    aggfunc="sum",
    fill_value=0
).reset_index()

zoning_area_wide.head(3)
````

````python
zoning_area_wide.columns.name = None
````

````python
zoning_area_wide.head(3)
````

````python
# Add total census tract area
tract_total_area = tract_poly[["GEOID", "geometry"]].copy()

tract_total_area["tract_total_area_km2"] = (
    tract_total_area.geometry.area / 1_000_000
)

tract_total_area = tract_total_area.drop(columns="geometry")


# Merge zoning area table with total tract area
zoning_share_wide = zoning_area_wide.merge(
    tract_total_area,
    on="GEOID",
    how="left",
    validate="one_to_one"
)
````

````python
# Convert zoning area to zoning share

# zoning_area_cols = []

# for c in zoning_area_wide.columns:
#     if c != "GEOID":
#         zoning_area_cols.append(c)

# Select raw zoning area columns besides GEOID column
# These columns are measured in km²
zoning_area_cols = [
    c for c in zoning_area_wide.columns
    if c != "GEOID"
]

# Convert zoning area to zoning share
for col in zoning_area_cols:
    zoning_share_wide[f"zoning_share_{col}"] = (
        zoning_share_wide[col] / zoning_share_wide["tract_total_area_km2"]
    )
````

````python
# The unit for those raw zoning area columns (downtown, mixed_use, residential, ...) are km². With this step, only selected GEOID and zoning_share_xxx variables will be merged back to master table
zoning_share_cols = [
    c for c in zoning_share_wide.columns
    if c.startswith("zoning_share_")
]

zoning_share_wide[["GEOID"] + zoning_share_cols].head()
````

````python
# Make sure GEOID format is consistent
tract_route_df["GEOID"] = tract_route_df["GEOID"].astype(str).str.zfill(11)
zoning_share_wide["GEOID"] = zoning_share_wide["GEOID"].astype(str).str.zfill(11)

# Merge zoning share variables into the master table
tract_built_env_df = tract_route_df.merge(
    zoning_share_wide[["GEOID"] + zoning_share_cols],
    on="GEOID",
    how="left",
    validate="one_to_one"
)

# Fill missing zoning share values with 0
tract_built_env_df[zoning_share_cols] = (
    tract_built_env_df[zoning_share_cols].fillna(0)
)

tract_built_env_df.head(3)
````

## 5. Master Table

````python
# Make sure GEOID format is consistent before merging
tract_route_df["GEOID"] = tract_route_df["GEOID"].astype(str).str.zfill(11)
zoning_share_wide["GEOID"] = tract_route_df["GEOID"].astype(str).str.zfill(11)
landuse_share_wide["GEOID"] = tract_route_df["GEOID"].astype(str).str.zfill(11)

# Select only final share variables
zoning_share_cols = [
    c for c in zoning_share_wide.columns
    if c.startswith("zoning_share_")
]

landuse_share_cols = [
    c for c in landuse_share_wide.columns
    if c.startswith("landuse_share_")
]
````

````python
# Merge zoning and landuse shares into the tract-level route/crash table
# Rebuild final master table from clean tract_route_df
tract_built_env_df = tract_route_df.copy()

tract_built_env_df = tract_built_env_df.merge(
    zoning_share_wide[["GEOID"] + zoning_share_cols],
    on="GEOID",
    how="left",
    validate="one_to_one"
)

tract_built_env_df = tract_built_env_df.merge(
    landuse_share_wide[["GEOID"] + landuse_share_cols],
    on="GEOID",
    how="left",
    validate="one_to_one"
)

# Fill only share variables
share_cols = zoning_share_cols + landuse_share_cols
tract_built_env_df[share_cols] = tract_built_env_df[share_cols].fillna(0)

tract_built_env_df.head(3)
````

## 6. EDA Comparison Logic: High-Crash vs. Lower-Crash Route-Exposed Tracts

In this EDA step, the analysis compares two groups of census tracts:

- **High-crash route-exposed tracts**
- **Lower-crash route-exposed tracts**

A census tract is considered **route-exposed** if it has major route exposure. Among those route-exposed tracts, the tracts with crash density in the top 25% are labeled as **high-crash route-exposed tracts**.

In this dataset:

- **52 GEOIDs** are classified as high-crash route-exposed tracts.
- **153 GEOIDs** are classified as lower-crash route-exposed tracts.

The goal is to compare the average zoning and land use composition between these two groups.

For example, for **mixed-use zoning share**:

- Among the 52 high-crash GEOIDs, what is their average mixed-use zoning share?
- Among the 153 lower-crash GEOIDs, what is their average mixed-use zoning share?

Example:

```text
High-crash tracts:

GEOID A: mixed_use share = 0.30
GEOID B: mixed_use share = 0.20
GEOID C: mixed_use share = 0.40

Average mixed-use share = 0.30

avg_landuse_commercial for the high-crash group
= the average commercial land use share across the 52 high-crash route-exposed tracts

````python
route_eda_df = tract_built_env_df[
    tract_built_env_df["route_exposed"]
].copy()

built_env_compare = (
    route_eda_df
    .groupby("high_crash_route_tract")
    .agg(
        n_tracts=("GEOID", "count"),
        avg_crash_density=("avg_annual_crash_density", "mean"),
        avg_major_route_density=("major_route_density_km_per_km2", "mean"),
        avg_major_route_share=("major_route_share", "mean"),
        avg_road_density=("road_density_km_per_km2", "mean"),
        avg_signal_density=("signal_density_km2", "mean"),
        avg_intersection_density=("intersection_density_km2", "mean"),

        avg_black_share=("black_alone_share", "mean"),
        avg_hispanic_share=("hispanic_share", "mean"),
        avg_white_share=("white_alone_share", "mean"),
        avg_asian_share=("asian_alone_share", "mean"),
        avg_two_or_more_races_share=("two_or_more_races_share", "mean"),
        avg_other_race_share=("other_race_share", "mean"),
        

        avg_zoning_mixed_use=("zoning_share_mixed_use", "mean"),
        avg_zoning_residential=("zoning_share_residential", "mean"),
        avg_zoning_residential_apt=("zoning_share_residential_apt", "mean"),
        avg_zoning_residential_flat=("zoning_share_residential_flat", "mean"),
        avg_zoning_pdr=("zoning_share_production_distribution_repair", "mean"),
        avg_zoning_unzoned=("zoning_share_unzoned", "mean"),

        avg_landuse_commercial=("landuse_share_commercial", "mean"),
        avg_landuse_residential=("landuse_share_residential", "mean"),
        avg_landuse_institutional=("landuse_share_institutional", "mean"),
        avg_landuse_vacant=("landuse_share_vacant", "mean"),
        avg_landuse_mixed=("landuse_share_mixed", "mean")
    )
    .round(3)
)

built_env_compare.index = built_env_compare.index.map({
    0: "Lower-crash route-exposed tracts",
    1: "Higher-crash route-exposed tracts"
})

# Remove the index name shown above the row labels
built_env_compare.index.name = None
````

### Initial EDA Table Interpretation

This table compares average tract-level characteristics between lower-crash and higher-crash route-exposed census tracts.

Each row represents one crash group, not one individual census tract. The values in the table are group averages across all GEOIDs in that group.

For example, `avg_landuse_commercial` represents the average commercial land use share among census tracts in each crash group. Similarly, `avg_zoning_mixed_use` represents the average mixed-use zoning share among census tracts in each crash group.

This table helps identify whether higher-crash route-exposed tracts tend to have different built-environment characteristics than lower-crash route-exposed tracts.
